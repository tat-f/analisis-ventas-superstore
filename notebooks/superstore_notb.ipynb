# Analisis de datos - proyecto de un Superstore

###Contexto

Con la creciente demanda y la competencia feroz en el mercado, un gigante de supertiendas busca tu conocimiento para entender qué es lo que mejor les funciona. Les gustaría entender qué productos, regiones, categorías y segmentos de clientes deberían dirigirse o evitar.

#####Preguntas de negocio:


1. ¿Qué categorías y subcategorías generan más ventas y cuáles dejan más beneficio?
2. ¿Hay productos que se venden mucho pero dan pérdida o margen bajo?
3. ¿Cómo afecta el descuento al beneficio? ¿Desde qué nivel de descuento empieza a haber pérdidas?
4. ¿Qué pocos productos concentran el 80% de las ventas? (análisis ABC)
5. ¿Cómo evolucionan las ventas por mes, y hay estacionalidad?

```python
#importamos librerias
import pandas as pd
```

```python
#cargo el dataset
df = pd.read_csv("/content/Superstore.csv", encoding='latin-1')
```

No hay presencia de valores nulos ni de ejemplos duplicados*

Se acomodan los tipos de datos, para que las fechas esten en 'date'

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
# Margen: beneficio como porcentaje de la venta
df['Margen'] = df['Profit'] / df['Sales']
df.describe()
```

Con **describe** podemos observar estadisticas generales de cada columna numerica.
Observamos que los datos van desde el año 2014 al 2017.
Una venta de 22.638 cuando la mediana es 54, y un beneficio de -6.600 contra una mediana de 8,67.
Hay pedidos que tienen hasta un 80% de descuento pero la mitad reciben un 20%.

*Row ID* y *Postal Code* no aportan valor.

```python
# Pregunta 1: ventas y beneficio por categoría y subcategoría
cat = df.groupby('Category')[['Sales', 'Profit']].sum()
cat['Margen %'] = cat['Profit'] / cat['Sales'] * 100
print(cat.sort_values('Profit', ascending=False).round(1))

sub = df.groupby(['Category', 'Sub-Category'])[['Sales', 'Profit']].sum()
sub['Margen %'] = sub['Profit'] / sub['Sales'] * 100
sub = sub.sort_values('Profit')
print(sub.round(1))
```

**Furniture** vende casi lo mismo que Technology y Office Supplies (742k), pero deja 18k de beneficio contra 145k y 122k.

**Subcategorías con pérdida** :
Son Tables (-17.725), Bookcases (-3.472) y Supplies (-1.189). Tienen buenas ventas pero no le ganan. ¿porque?

Labels tiene 44% de margen pero vendió solo 12k, así que deja 5,5k. Es muy rentable pero pequeño: aunque se venda mucho más. Copiers combina las dos cosas (37% de margen y el mayor beneficio) y Paper también (43% de margen y 34k de beneficio). Esas serían las categorías a cuidar y potenciar.

Machines vende 189k con solo 1,8% de margen. Casi no deja nada.

```python
# Pregunta 2: productos que dan pérdida
prod = df.groupby('Product Name')[['Sales', 'Profit']].sum()
perdida = prod[prod['Profit'] < 0].sort_values('Profit')
print("Productos con pérdida total:", len(perdida), "de", len(prod))
print(perdida.head(10).round(1))
```

aca observamos que de los 1850 productos totales, 301 son los que generan perdida y conviene verlos!

```python
# Pregunta 3: efecto del descuento sobre el beneficio
df['Rango descuento'] = pd.cut(
    df['Discount'],
    bins=[-0.01, 0, 0.2, 0.4, 1],
    labels=['Sin descuento', '1-20%', '21-40%', 'Más de 40%']
)
desc = df.groupby('Rango descuento', observed=True)[['Sales', 'Profit']].sum()
desc['Margen %'] = desc['Profit'] / desc['Sales'] * 100
print(desc.round(1))
```

Se obseva que a partir de mas del 20% de descuento, se generan perdidas.

```python
# Descuento promedio por subcategoría, junto con su margen
sub2 = df.groupby('Sub-Category').agg(
    Ventas=('Sales', 'sum'),
    Beneficio=('Profit', 'sum'),
    Desc_prom=('Discount', 'mean')
)
sub2['Margen %'] = sub2['Beneficio'] / sub2['Ventas'] * 100
print(sub2.sort_values('Margen %').round(2))
```

```python
# Tables, Bookcases, Supplies: ¿qué pasa con y sin descuento?
foco = df[df['Sub-Category'].isin(['Tables', 'Bookcases', 'Supplies'])]
res = foco.groupby(['Sub-Category', 'Rango descuento'], observed=True)[['Sales', 'Profit']].sum()
res['Margen %'] = res['Profit'] / res['Sales'] * 100
print(res.round(1))
```

Vemos que, en bookcases hasta un 20% se podria llegar (no conviene tanto) pero aun asi algo de ganancia nos deja.
En supplies y tables, directamente no conviene hacer descuentos. Por ende se asume que la causa de perdida en estos productos es debido al descuento aplicado.

```python
# ¿Cuánto pierden en total los 301 y qué tan concentrada está la pérdida?
print("Pérdida total de esos productos:", round(perdida['Profit'].sum(), 1))
print("Top 10 explican:", round(perdida['Profit'].head(10).sum() / perdida['Profit'].sum() * 100, 1), "% de esa pérdida")

# Descuento promedio de los 10 productos que más pierden
top10 = perdida.head(10).index
print(df[df['Product Name'].isin(top10)]
      .groupby('Product Name')['Discount'].mean().round(2))
```

```python
# Pregunta 3 completa: efecto del descuento en todo el catálogo
desc = df.groupby('Rango descuento', observed=True)[['Sales', 'Profit']].sum()
desc['Margen %'] = desc['Profit'] / desc['Sales'] * 100
print(desc.round(1))
```

```python
# Pregunta 4: análisis ABC por producto
abc = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).to_frame()
abc['% acumulado'] = abc['Sales'].cumsum() / abc['Sales'].sum() * 100
abc['Clase'] = pd.cut(abc['% acumulado'], bins=[0, 80, 95, 100.01], labels=['A', 'B', 'C'])

resumen = abc.groupby('Clase', observed=True).agg(
    Productos=('Sales', 'count'),
    Ventas=('Sales', 'sum')
)
resumen['% productos'] = resumen['Productos'] / len(abc) * 100
resumen['% ventas'] = resumen['Ventas'] / abc['Sales'].sum() * 100
print(resumen.round(1))
```

```python
# Pregunta 5: evolución mensual de ventas y beneficio
mensual = df.groupby(df['Order Date'].dt.to_period('M'))[['Sales', 'Profit']].sum()
print(mensual.round(0))

# Estacionalidad: promedio por mes del año
estac = df.groupby(df['Order Date'].dt.month)[['Sales']].sum() / 4
print(estac.round(0))
```
